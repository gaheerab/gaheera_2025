## Popcorn Hack #2
### Use the same code from the Rock, Paper, Scissors simulation. Run the cell in your notebook and when prompted to, type out your choice of the three moves. Share your output with the class

```python
import random

def play_rock_paper_scissors():
    choices = ['rock', 'paper', 'scissors']
    computer_choice = random.choice(choices)
    user_choice = input("Enter your choice (rock, paper, or scissors): ")

    if user_choice not in choices:
        print("Invalid choice. Please try again.")
        return

    print("Computer chose:", computer_choice)
    print("You chose:", user_choice)

    if user_choice == computer_choice:
        print("It's a tie!")
    elif (user_choice == 'rock' and computer_choice == 'scissors') or (user_choice == 'paper' and computer_choice == 'rock') or (user_choice == 'scissors' and computer_choice == 'paper'):
        print("You win!")
    else:
        print("You lose!")

play_rock_paper_scissors()
```

## Homework Hack 1: Random Team Assignment
### Objective: Randomly assign students to teams for a group project.

Task:

You are given a list of 15 students.
The goal is to randomly assign each student to one of 3 teams (be creative with the names)
Implement the random assignment:
Randomly assign each student to a team
Print out the list of students and their assigned teams.


```python
import random

# List of 15 students
students = [
    "Aanya", "Brandon", "Chloe", "Diego", "Emma",
    "Farah", "George", "Hana", "Isaac", "Jasmine",
    "Kiran", "Liam", "Maya", "Nathan", "Olivia"
]

# Creative team names
teams = ["Fire Falcons", "Aqua Knights", "Storm Seekers"]

# Shuffle the list of students randomly
random.shuffle(students)

# Create a dictionary to hold the team assignments
team_assignments = {team: [] for team in teams}

# Distribute students evenly into the teams
for i, student in enumerate(students):
    team = teams[i % len(teams)]  # This ensures the students are distributed evenly
    team_assignments[team].append(student)

# Print out the results
for team, members in team_assignments.items():
    print(f"\nTeam {team} ({len(members)} members):")
    for member in members:
        print(f"  - {member}")
```

## Hack 2: Random Weather
### Objective: Generate random weather patterns over a set number of days.

Task:

You’re simulating the weather for a 7-day forecast.
The weather each day can be one of three options: Sunny, Cloudy, or Rainy.
Write a program randomly selecting the weather type for each of the 7 days.
Print the weather for each day

```python
import random

# List of possible weather conditions
weather_options = ["Sunny", "Cloudy", "Rainy"]

# Simulate the weather for the next 7 days
forecast = {}

for day in range(1, 8):  # 7 days forecast
    weather = random.choice(weather_options)
    forecast[f"Day {day}"] = weather

# Print the forecast for each day
print("7-Day Weather Forecast:")
for day, weather in forecast.items():
    print(f"{day}: {weather}")
```

## Homework Hack: Simulate a Random Coffee Shop Queue
### Objective: Simulate a random coffee shop queue and determine how long it takes for customers to get served.

Task: Simulate a coffee shop with 5 customers. Each customer has a random service time (between 1 and 5 minutes). The coffee shop serves customers in the order they arrive. Calculate and print the total time it takes for all 5 customers to be served.

```python
import random

# Number of customers
num_customers = 5

# Generate a list of random service times (between 1 and 5 minutes)
service_times = [random.randint(1, 5) for _ in range(num_customers)]

# Simulate the queue and calculate the total time
total_time = 0

print("Coffee Shop Queue Simulation:")
for i, time in enumerate(service_times, 1):
    total_time += time
    print(f"Customer {i} is served in {time} minutes.")

# Print the total time taken to serve all customers
print(f"\nTotal time to serve all customers: {total_time} minutes.")
```
